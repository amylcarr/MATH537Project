```python
import pandas as pd

# Load the Excel file
file_path = "C:/Users/amyyy/OneDrive/Desktop/MATH 537/FINALPROJ/Housing/ACSDT5Y2023.B25014-Data.csv"  # Update with the correct file path
df = pd.read_csv(file_path, dtype=str)

# Define relevant columns
columns_to_keep = ["GEO_ID", "NAME", "B25014_001E", "B25014_002E", "B25014_003E", 
                   "B25014_004E", "B25014_005E", "B25014_006E", "B25014_007E", 
                   "B25014_008E", "B25014_009E", "B25014_010E", "B25014_011E", 
                   "B25014_012E", "B25014_013E"]

# Convert relevant columns to numeric (for summing)
for col in columns_to_keep[2:]:  # Skip GEO_ID and NAME
    df[col] = pd.to_numeric(df[col], errors="coerce")  # Convert to numbers, set errors to NaN

def extract_county(name):
    parts = name.split(", ")
    if len(parts) >= 3:
        return f"{parts[-2]}, {parts[-1]}"  # "Alamance County, North Carolina"
    return name  # Fallback if format is unexpected

df["County"] = df["NAME"].apply(lambda x: extract_county(x) if pd.notna(x) else "")

# Group by County and sum the numeric columns
df_grouped = df.groupby("County")[columns_to_keep[2:]].sum().reset_index()

# Rename columns for clarity
df_grouped.rename(columns={
    "B25014_001E": "Total Households",
    "B25014_002E": "Owner-occupied: Total",
    "B25014_003E": "Owner-occupied: 1.00 or fewer occupants per room",
    "B25014_004E": "Owner-occupied: 1.01 to 1.50 occupants per room",
    "B25014_005E": "Owner-occupied: 1.51 or more occupants per room",
    "B25014_006E": "Renter-occupied: Total",
    "B25014_007E": "Renter-occupied: 1.00 or fewer occupants per room",
    "B25014_008E": "Renter-occupied: 1.01 to 1.50 occupants per room",
    "B25014_009E": "Renter-occupied: 1.51 or more occupants per room",
    "B25014_010E": "Owner-occupied: 1.01 to 1.50 occupants per room (Percentage)",
    "B25014_011E": "Owner-occupied: 1.51 or more occupants per room (Percentage)",
    "B25014_012E": "Renter-occupied: 1.01 to 1.50 occupants per room (Percentage)",
    "B25014_013E": "Renter-occupied: 1.51 or more occupants per room (Percentage)"
}, inplace=True)

# Save the cleaned and aggregated data
df_selected.to_csv("C:/Users/amyyy/OneDrive/Desktop/MATH 537/FINALPROJ/Housing/housing_data_combined_county.csv", index=False)

print("Cleaned and aggregated data saved as 'cleaned_housing_data_by_county.xlsx'.")
```
