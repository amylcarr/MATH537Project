Merge all SDOH data into one file. Inner join on GEOID and County Name columns

```python
import pandas as pd

files = [
    "cleaned_education_data.csv",
    "cleaned_housing_data.csv",
    "cleaned_employment_data.csv",
    "cleaned_household_income_data.csv",
    "cleaned_insurance_data.csv"
]

data_frames = [pd.read_csv(f"C:/Users/amyyy/OneDrive/Desktop/MATH_537/FINALPROJ/Data/{file}") for file in files]

merged_df = data_frames[0]

for df in data_frames[1:]:
    merged_df = pd.merge(merged_df, df, how='inner', on=["GEOID", "County Name"])

merged_df.to_csv("C:/Users/amyyy/OneDrive/Desktop/MATH_537/FINALPROJ/Data/merged_sdoh_data.csv")
```

We now extract specifically the county GEOID for each row

```python
df = pd.read_csv("C:/Users/amyyy/OneDrive/Desktop/MATH_537/FINALPROJ/Data/merged_sdoh_data.csv")

df['county_geoid'] = df['GEOID'].str.slice(9, 14).radd('US')

df.to_csv("C:/Users/amyyy/OneDrive/Desktop/MATH_537/FINALPROJ/Data/merged_sdoh_data1.csv")
```
