```python
import pandas as pd
import glob

folder_path = "C:/Users/amyyy/OneDrive/Desktop/MATH_537/FINALPROJ/Data/Model/"
file_paths = glob.glob(folder_path + "*ModelData.csv")

all_years_df = pd.concat([pd.read_csv(file) for file in file_paths], ignore_index=True)
all_years_df = all_years_df.dropna(subset=[
    'Renter-occupied: 1.51 or more occupants per room',
    'County HIV Cases',
    'County HIV Rate',
    'Crude Rate',
    'Population',
    'Rate: Health Insurance',
    'Rate: No Health Insurance',
    'Percent Less Than HS',
    'Percent Some College',
    'Percent Unemployed',
    'Percent Owner-Occupied Crowded',
    'Median Household Income'
])
all_years_df['Deaths'] = pd.to_numeric(all_years_df['Deaths'], errors='coerce')
all_years_df['Crude Rate'] = pd.to_numeric(all_years_df['Crude Rate'], errors='coerce')
all_years_df = all_years_df.dropna(subset=['Crude Rate'])

agg_df = all_years_df.groupby('County Code').agg({
    'Renter-occupied: 1.51 or more occupants per room': 'mean',
    'County HIV Cases': 'mean',
    'County HIV Rate': 'mean',
    'Deaths': 'mean',
    'Crude Rate': 'mean',
    'Population': 'mean',
    'Rate: Health Insurance': 'mean',
    'Rate: No Health Insurance': 'mean',
    'Percent Less Than HS': 'mean',
    'Percent Some College': 'mean',
    'Percent Unemployed': 'mean',
    'Percent Owner-Occupied Crowded': 'mean',
    'Median Household Income': 'mean'
}).reset_index()
```
